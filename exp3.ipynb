```python
import pandas as pd

df = pd.read_csv("/Users/isirirram/DAV/house price/For_EDA_dataset.csv")
print(df.shape)
df.head()
```

```python
df_mean = df.copy()
num_cols = df_mean.select_dtypes(include='number').columns

for col in num_cols:
    mean_val = df[col].mean()
    print("Mean", col, ":", mean_val)
    df_mean[col] = df_mean[col].fillna(mean_val)

print(df_mean.isnull().sum())
df_mean.head()
```

```python
df_median = df.copy()

for col in num_cols:
    df_median[col] = df_median[col].fillna(df[col].median())

print(df_median.isnull().sum())
df_median.head()
```

```python
df_ffill = df.copy()
cat_cols = df_ffill.select_dtypes(exclude='number').columns

print("Before Forward Fill:")
print(df_ffill[cat_cols].isnull().sum())

df_ffill[cat_cols] = df_ffill[cat_cols].ffill()

print("\nAfter Forward Fill:")
print(df_ffill[cat_cols].isnull().sum())
```

```python
from sklearn.preprocessing import LabelEncoder

df_label = df_ffill.copy()

for col in cat_cols:
    encoder = LabelEncoder()
    df_label[col + '_Encoded'] = encoder.fit_transform(df_label[col])
    print(col, encoder.classes_)

df_label.head()
```
